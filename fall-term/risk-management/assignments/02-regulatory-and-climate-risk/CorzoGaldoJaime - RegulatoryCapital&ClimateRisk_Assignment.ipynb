## A. Balance Sheet Dynamics and Risk-Weighted Assets

A bank's balance sheet satisfies the accounting identity

$$
A = L + E
$$

where:

- $A$ denotes total assets,
- $L$ denotes total liabilities,
- $E$ denotes equity.

Equivalently,

$$
E = A - L
$$

A simple measure of bank capitalization is the equity-to-assets ratio:

$$
\frac{E}{A}
$$

A higher ratio means that a larger proportion of the bank's assets is financed by equity rather than debt or deposits.

However, this measure treats all assets as equally risky. Regulatory capital frameworks therefore introduce **risk-weighted assets (RWA)**, where each exposure is weighted according to its risk:

$$
RWA = \sum_i A_i w_i
$$

where $A_i$ is the exposure to asset $i$ and $w_i$ is its regulatory risk weight.

### Question c

A simple equity-to-assets ratio treats all assets as equally risky. Basel I instead assigns different regulatory risk weights to different asset classes.

The relevant risk weights for the final balance sheet are:

| Asset | Exposure ($MM) | Risk Weight |
|---|---:|---:|
| Cash | 0.5 | 0% |
| Mortgages | 2.0 | 50% |
| Corporate loan | 0.7 | 100% |
| Securitizations | 0.5 | 20% |
| Canadian government securities | 0.2 | 0% |

Risk-weighted assets are calculated as

$$
RWA = \sum_i A_i w_i
$$

Therefore,

$$
RWA
=
(0.5)(0)
+
(2.0)(0.50)
+
(0.7)(1.00)
+
(0.5)(0.20)
+
(0.2)(0)
$$

which gives

$$
RWA
=
1.0+0.7+0.1
=
1.8
$$

million dollars.

The corresponding equity-to-RWA ratio is

$$
\frac{E}{RWA}
=
\frac{0.9}{1.8}
=
50.0\%
$$

Thus, although the bank's simple equity-to-assets ratio is only

$$
\frac{E}{A}=23.08\%,
$$

its equity-to-RWA ratio is

$$
\frac{E}{RWA}=50.0\%.
$$

The difference arises because not all assets receive a 100% risk weight. In particular, cash and Canadian government securities receive a zero risk weight, mortgages receive a 50% weight, and securitizations receive a 20% weight.

This illustrates the motivation for regulatory RWA: the denominator of the capital ratio reflects the risk composition of the bank's assets rather than only their accounting value.

```python
# Final balance sheet after Step 6 ($MM)

cash = 0.5
mortgages = 2.0
corporate_loan = 0.7
securitizations = 0.5
government_securities = 0.2

liabilities = 3.0
equity = 0.9

total_assets = (
    cash
    + mortgages
    + corporate_loan
    + securitizations
    + government_securities
)

equity_to_assets = equity / total_assets

# Basel I risk weights
rw_cash = 0.00
rw_mortgages = 0.50
rw_corporate = 1.00
rw_securitizations = 0.20
rw_government = 0.00

rwa = (
    cash * rw_cash
    + mortgages * rw_mortgages
    + corporate_loan * rw_corporate
    + securitizations * rw_securitizations
    + government_securities * rw_government
)

equity_to_rwa = equity / rwa

print(f"Total assets: ${total_assets:.2f}MM")
print(f"Equity / Assets: {equity_to_assets:.2%}")
print(f"RWA: ${rwa:.2f}MM")
print(f"Equity / RWA: {equity_to_rwa:.2%}")
```

## B. RWA Using OSFI Scenario Parameters

The second part of the assignment studies how climate-related changes in credit risk could affect regulatory Risk-Weighted Assets (RWA).

OSFI's climate scenario methodology modifies the credit-risk parameters of a borrower through the following sequence:

$$
PD
\rightarrow
CPD
\rightarrow
climateCPD
\rightarrow
climatePD
\rightarrow
climateLGD
$$

The assignment assumes that these climate-adjusted credit-risk parameters can be inserted into the Basel RWA formula.

This is a simplifying assumption for the purpose of the exercise. The OSFI climate adjustments were originally designed for provisioning and Expected Credit Loss (ECL), rather than directly for regulatory RWA.

### Credit-Risk Parameters

Before introducing the climate adjustment, it is useful to distinguish the main credit-risk parameters used in the exercise.

#### Probability of Default (PD)

The Probability of Default measures the unconditional probability that a borrower defaults during a given period.

#### Conditional Probability of Default (CPD)

The Conditional Probability of Default measures the probability that a borrower defaults during a given period **conditional on having survived until the beginning of that period**.

For the first year,

$$
PD_1 = CPD_1
$$

because the borrower is assumed to be alive at the beginning of the analysis.

For later periods, the unconditional default probability must also account for survival through all previous periods. Therefore,

$$
PD_i
=
\left[
\prod_{j=1}^{i-1}
(1-CPD_j)
\right]
CPD_i
$$

#### Loss Given Default (LGD)

LGD measures the fraction of the exposure that is lost if default occurs.

For example,

$$
LGD = 40\%
$$

means that the bank expects to lose 40% of its exposure if the borrower defaults.

#### Exposure at Default (EAD)

EAD is the amount to which the bank is exposed at the time of default.

In this exercise,

$$
EAD = \$1,000,000
$$

### OSFI Climate Adjustment

OSFI introduces a climate add-on parameter, denoted by $c$ in this assignment, which modifies the borrower's conditional probability of default.

The climate-adjusted CPD is

$$
climateCPD_i
=
\frac{1}
{
1+\exp
\left[
-
\left(
\operatorname{logit}(CPD_i)+c
\right)
\right]
}
$$

where

$$
\operatorname{logit}(p)
=
\ln
\left(
\frac{p}{1-p}
\right)
$$

The transformation applies the climate adjustment to the **log-odds** of default rather than adding it directly to the probability.

Once the adjusted conditional probability is obtained, the corresponding unconditional climate-adjusted PD is

$$
climatePD_i
=
\left[
\prod_{j=1}^{i-1}
(1-climateCPD_j)
\right]
climateCPD_i
$$

Finally, OSFI uses the Frye-Jacobs relationship to obtain the climate-adjusted LGD:

$$
climateLGD_i
=
\frac{
\Phi
\left[
\Phi^{-1}(climatePD_i)
-
\Phi^{-1}(PD_i)
+
\Phi^{-1}(PD_i LGD_i)
\right]
}
{climatePD_i}
$$

where $\Phi$ denotes the standard normal cumulative distribution function and $\Phi^{-1}$ its inverse.

### Question d

We want to show that setting the climate add-on equal to zero leaves the original credit-risk parameters unchanged.

Consider first the climate-adjusted CPD:

$$
climateCPD_i
=
\frac{1}
{
1+\exp
\left[
-
\left(
\operatorname{logit}(CPD_i)+c
\right)
\right]
}
$$

Setting

$$
c=0
$$

gives

$$
climateCPD_i
=
\frac{1}
{
1+\exp
\left[
-\operatorname{logit}(CPD_i)
\right]
}
$$

Since the logistic function is the inverse of the logit function,

$$
\operatorname{logistic}
\left(
\operatorname{logit}(p)
\right)
=
p
$$

and therefore

$$
\boxed{
climateCPD_i = CPD_i
}
$$

If all conditional default probabilities remain unchanged, the corresponding unconditional default probabilities must also remain unchanged.

The climate-adjusted PD is

$$
climatePD_i
=
\left[
\prod_{j=1}^{i-1}
(1-climateCPD_j)
\right]
climateCPD_i
$$

Since

$$
climateCPD_j = CPD_j
$$

for every period $j$, it follows that

$$
climatePD_i
=
\left[
\prod_{j=1}^{i-1}
(1-CPD_j)
\right]
CPD_i
=
PD_i
$$

and therefore

$$
\boxed{
climatePD_i = PD_i
}
$$

Finally, consider the Frye-Jacobs relationship:

$$
climateLGD_i
=
\frac{
\Phi
\left[
\Phi^{-1}(climatePD_i)
-
\Phi^{-1}(PD_i)
+
\Phi^{-1}(PD_i LGD_i)
\right]
}
{climatePD_i}
$$

When $c=0$, we have already shown that

$$
climatePD_i = PD_i
$$

so the first two terms inside the normal CDF cancel:

$$
\Phi^{-1}(climatePD_i)
-
\Phi^{-1}(PD_i)
=
0
$$

Therefore,

$$
climateLGD_i
=
\frac{
\Phi
\left[
\Phi^{-1}(PD_i LGD_i)
\right]
}
{PD_i}
$$

Since $\Phi$ and $\Phi^{-1}$ are inverse functions,

$$
\Phi
\left[
\Phi^{-1}(PD_i LGD_i)
\right]
=
PD_i LGD_i
$$

and hence

$$
climateLGD_i
=
\frac{PD_i LGD_i}{PD_i}
=
LGD_i
$$

Therefore,

$$
\boxed{
climateLGD_i = LGD_i
}
$$

Thus, when the climate add-on is zero,

$$
\boxed{
climateCPD = CPD,
\qquad
climatePD = PD,
\qquad
climateLGD = LGD
}
$$

This is an important consistency property of the parameterization: if there is no climate adjustment, the model exactly reproduces the original credit-risk parameters.

### Question e

Consider a loan with the following parameters:

$$
CPD = 1\%
$$

$$
LGD = 40\%
$$

$$
EAD = \$1,000,000
$$

The assignment additionally specifies

$$
M=1
$$

and

$$
R=0.15
$$

where $M$ is maturity and $R$ is the asset-correlation parameter.

Because regulatory capital considers default risk over the next year, the one-year PD and CPD coincide:

$$
PD = CPD = 0.01
$$

The Basel IRB formula used in the lecture is

$$
RWA
=
12.5
\times
EAD
\times
K
$$

where

$$
K
=
\left[
LGD
\,
\Phi
\left(
\frac{
\Phi^{-1}(PD)
+
\sqrt{R}\Phi^{-1}(0.999)
}
{
\sqrt{1-R}
}
\right)
-
LGD \times PD
\right]
\times
\frac{
1+(M-2.5)b
}{
1-1.5b
}
$$

and

$$
b
=
\left(
0.11852
-
0.05478\ln(PD)
\right)^2
$$

Since $M=1$, the maturity adjustment simplifies to

$$
\frac{
1+(M-2.5)b
}{
1-1.5b
}
=
\frac{
1-1.5b
}{
1-1.5b
}
=
1
$$

Therefore, for this exercise,

$$
K
=
LGD
\,
\Phi
\left(
\frac{
\Phi^{-1}(PD)
+
\sqrt{R}\Phi^{-1}(0.999)
}
{
\sqrt{1-R}
}
\right)
-
LGD \times PD
$$

Using

$$
PD=0.01,
\qquad
LGD=0.40,
\qquad
R=0.15
$$

gives approximately

$$
K = 0.040106
$$

and therefore

$$
RWA
=
12.5
\times
1{,}000{,}000
\times
0.040106
$$

$$
\boxed{
RWA \approx \$501{,}324
}
$$

Thus, the base RWA associated with the $1$ million exposure is approximately $0.501$ million.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import norm
```

```python
def calculate_rwa(pd, lgd, ead, maturity=1.0, correlation=0.15):
    """
    Basel IRB RWA for a wholesale credit exposure.
    """

    b = (0.11852 - 0.05478 * np.log(pd)) ** 2

    maturity_adjustment = (
        1 + (maturity - 2.5) * b
    ) / (
        1 - 1.5 * b
    )

    stressed_default_probability = norm.cdf(
        (
            norm.ppf(pd)
            + np.sqrt(correlation) * norm.ppf(0.999)
        )
        / np.sqrt(1 - correlation)
    )

    K = (
        lgd * stressed_default_probability
        - lgd * pd
    ) * maturity_adjustment

    rwa = 12.5 * ead * K

    return rwa
```

```python
# Base loan parameters

base_cpd = 0.01
base_pd = base_cpd
base_lgd = 0.40
ead = 1_000_000

M = 1.0
R = 0.15

base_rwa = calculate_rwa(
    pd=base_pd,
    lgd=base_lgd,
    ead=ead,
    maturity=M,
    correlation=R
)

print(f"Base RWA: ${base_rwa:,.2f}")
```

```python
def logit(p):
    return np.log(p / (1 - p))


def logistic(x):
    return 1 / (1 + np.exp(-x))


def calculate_climate_cpd(cpd, c):
    """
    Apply the OSFI climate add-on to conditional PD.
    """

    return logistic(
        logit(cpd) + c
    )
```

```python
def calculate_unconditional_pd(cpd_series):
    """
    Convert a sequence of conditional PDs into unconditional PDs.
    """

    cpd_series = np.asarray(cpd_series)

    survival_probability = 1.0
    pd_series = []

    for cpd in cpd_series:
        pd = survival_probability * cpd
        pd_series.append(pd)

        survival_probability *= (1 - cpd)

    return np.array(pd_series)
```

```python
def calculate_climate_lgd(base_pd, base_lgd, climate_pd):
    """
    Frye-Jacobs climate LGD adjustment.
    """

    numerator = norm.cdf(
        norm.ppf(climate_pd)
        - norm.ppf(base_pd)
        + norm.ppf(base_pd * base_lgd)
    )

    return numerator / climate_pd
```

```python
# Sanity check: c = 0 should leave all parameters unchanged

c = 0.0

climate_cpd = calculate_climate_cpd(base_cpd, c)

# Over the first year, conditional and unconditional PD coincide
climate_pd = climate_cpd

climate_lgd = calculate_climate_lgd(
    base_pd=base_pd,
    base_lgd=base_lgd,
    climate_pd=climate_pd
)

print(f"Base CPD:       {base_cpd:.4%}")
print(f"Climate CPD:    {climate_cpd:.4%}")
print()
print(f"Base PD:        {base_pd:.4%}")
print(f"Climate PD:     {climate_pd:.4%}")
print()
print(f"Base LGD:       {base_lgd:.4%}")
print(f"Climate LGD:    {climate_lgd:.4%}")
```

The next step is to study how the climate adjustment changes RWA.

The climate add-on is allowed to vary over

$$
0 \leq c \leq 2
$$

For each value of $c$, the calculation proceeds as follows:

$$
CPD
\rightarrow
climateCPD
\rightarrow
climatePD
\rightarrow
climateLGD
\rightarrow
climateRWA
$$

For regulatory RWA, the assignment specifically requires the use of $climateCPD$, rather than $climatePD$, as the default-probability input to the Basel formula.

The quantity of interest is

$$
\frac{climateRWA}{baseRWA}
$$

A value of $1$ means that the climate adjustment has no effect on RWA, while a value above $1$ represents an increase relative to the baseline.

```python
c_values = np.linspace(0, 2, 201)

results = []

for c in c_values:

    # Step 1: climate-adjusted CPD
    climate_cpd = calculate_climate_cpd(
        base_cpd,
        c
    )

    # Step 2: over the first year, PD = CPD
    climate_pd = climate_cpd

    # Step 3: climate-adjusted LGD
    climate_lgd = calculate_climate_lgd(
        base_pd=base_pd,
        base_lgd=base_lgd,
        climate_pd=climate_pd
    )

    # Step 4: climate RWA
    # The assignment requires climateCPD, not climatePD,
    # as the PD input to the regulatory RWA formula.
    climate_rwa = calculate_rwa(
        pd=climate_cpd,
        lgd=climate_lgd,
        ead=ead,
        maturity=M,
        correlation=R
    )

    rwa_ratio = climate_rwa / base_rwa

    results.append(
        {
            "c": c,
            "Climate CPD": climate_cpd,
            "Climate PD": climate_pd,
            "Climate LGD": climate_lgd,
            "Climate RWA": climate_rwa,
            "RWA Ratio": rwa_ratio
        }
    )

results = pd.DataFrame(results)
```

```python
selected_c = [0.0, 0.5, 1.0, 1.5, 2.0]

summary = results[
    results["c"].round(2).isin(selected_c)
].copy()

summary["Climate CPD"] = summary["Climate CPD"].map(
    lambda x: f"{x:.2%}"
)

summary["Climate LGD"] = summary["Climate LGD"].map(
    lambda x: f"{x:.2%}"
)

summary["Climate RWA"] = summary["Climate RWA"].map(
    lambda x: f"${x:,.0f}"
)

summary["RWA Ratio"] = summary["RWA Ratio"].map(
    lambda x: f"{x:.2f}x"
)

summary
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    results["c"],
    results["RWA Ratio"],
    linewidth=2
)

plt.axhline(
    y=1,
    linestyle="--",
    linewidth=1
)

plt.xlabel("Climate Add-on, c")
plt.ylabel("Climate RWA / Base RWA")
plt.title("Impact of the OSFI Climate Add-on on Risk-Weighted Assets")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

### Question f

The results show that OSFI's climate parameterization has the potential to produce a substantial increase in RWA.

The climate add-on does not directly add a fixed amount to the probability of default. Instead, it shifts the log-odds of the conditional default probability:

$$
\operatorname{logit}(climateCPD)
=
\operatorname{logit}(CPD)+c
$$

As a result, the relationship between $c$ and the final RWA is nonlinear.

There are also two reinforcing channels through which the climate adjustment affects RWA.

First, a positive value of $c$ increases $climateCPD$. Since default probability is an input into the Basel IRB capital formula, this directly changes the credit-risk capital requirement.

Second, the Frye-Jacobs relationship links the deterioration in PD to an increase in LGD. Therefore, the climate adjustment simultaneously increases both the probability that a default occurs and the proportion of the exposure that is lost conditional on default.

For the loan considered here, the effect is economically significant. The base RWA is approximately

$$
\$0.501\text{ million}
$$

while at

$$
c=2
$$

the climate-adjusted RWA is approximately

$$
\$1.980\text{ million}
$$

or almost four times the original amount:

$$
\frac{climateRWA}{baseRWA}
\approx
3.95
$$

This demonstrates that OSFI's parameterization is capable of translating a climate-risk shock into a large change in credit-risk capital when the adjusted PD and LGD are applied to the Basel RWA framework.

It is important, however, to interpret this result within the assumptions of the exercise. The assignment deliberately applies OSFI's climate-adjusted PD and LGD parameters to the RWA formula, even though the parameters were originally introduced for provisioning and Expected Credit Loss calculations.
